# Testing PlotKit Suggestions API

This notebook demonstrates and tests the `recommender()` function from PlotKit.

## Features Tested
- Visualization recommendations based on data
- Multi-provider support for recommendations
- Model selection with `selected_models`
- Interactive and non-interactive modes
- Error handling

## Setup: Import Libraries

```python
import os
import pandas as pd
from dotenv import load_dotenv
from plotsense.visual_suggestion.suggestions import recommender

# Load environment variables from .env file
load_dotenv()

print("✅ Libraries imported successfully")
```

## Step 1: Create Sample Data

```python
# Create a sample DataFrame for testing
data = {
    "Year": [2020, 2021, 2022, 2023, 2024],
    "Sales": [150000, 200000, 250000, 300000, 350000],
    "Profit": [40000, 50000, 65000, 80000, 95000],
    "Customers": [100, 150, 200, 250, 300],
    "Region": ["North", "South", "East", "West", "Central"]
}

df = pd.DataFrame(data)
print("✅ Sample data created:")
print(df)
print(f"\nShape: {df.shape}")
print(f"Columns: {list(df.columns)}")
```

## Step 2: Setup API Keys

```python
# Option 1: Load from environment variables (recommended)
api_keys = {}

# Only include keys that are actually set in environment
if os.getenv("GROQ_API_KEY"):
    api_keys["groq"] = os.getenv("GROQ_API_KEY")

if os.getenv("OPENAI_API_KEY"):
    api_keys["openai"] = os.getenv("OPENAI_API_KEY")

if os.getenv("ANTHROPIC_API_KEY"):
    api_keys["anthropic"] = os.getenv("ANTHROPIC_API_KEY")

print(f"Available API keys: {list(api_keys.keys())}")

# Option 2: Pass keys directly (less secure, for testing only)
# api_keys = {"groq": "your-key-here"}

# Option 3: Leave empty for interactive mode
# api_keys = {}  # Will prompt for keys interactively
```

## Step 3: Test 1 - Basic Recommendations (Default)

```python
# Get visualization recommendations
try:
    recommendations = recommender(
        df,
        n=3,  # Number of recommendations
        api_keys=api_keys,
        interactive=len(api_keys) == 0,
        timeout=30
    )
    print("✅ Recommendations generated:")
    print(f"\n{recommendations}")
except Exception as e:
    print(f"❌ Error: {e}")
    print("\nNote: You may need to:")
    print("1. Set API keys in environment variables")
    print("2. Or pass interactive=True to be prompted for keys")
```

## Step 4: Test 2 - Recommendations with Specific Provider

```python
# Get recommendations using OpenAI
try:
    recommendations = recommender(
        df,
        n=2,
        api_keys=api_keys,
        selected_models=[("openai", "gpt-4o-mini")],
        interactive=len(api_keys) == 0,
        timeout=30
    )
    print("✅ OpenAI recommendations generated:")
    print(f"\n{recommendations}")
except Exception as e:
    print(f"❌ Error: {e}")
```

## Step 5: Test 3 - Recommendations with Groq

```python
# Get recommendations using Groq (faster, lower cost)
try:
    recommendations = recommender(
        df,
        n=2,
        api_keys=api_keys,
        selected_models=[("groq", "llama-3.1-8b-instant")],
        interactive=len(api_keys) == 0,
        timeout=30
    )
    print("✅ Groq recommendations generated:")
    print(f"\n{recommendations}")
except Exception as e:
    print(f"❌ Error: {e}")
```

## Step 6: Test 4 - Different Data Types

```python
# Create a different dataset - with categories
sales_by_category = pd.DataFrame({
    "Product": ["Laptops", "Phones", "Tablets", "Accessories"],
    "Q1": [50000, 80000, 30000, 20000],
    "Q2": [60000, 90000, 35000, 25000],
    "Q3": [70000, 100000, 40000, 30000],
    "Q4": [90000, 120000, 50000, 40000]
})

print("Sample data for category analysis:")
print(sales_by_category)

# Get recommendations for this data
try:
    recommendations = recommender(
        sales_by_category,
        n=3,
        api_keys=api_keys,
        interactive=len(api_keys) == 0,
        timeout=30
    )
    print("\n✅ Recommendations for category data:")
    print(f"\n{recommendations}")
except Exception as e:
    print(f"\n❌ Error: {e}")
```

## Step 7: Supported Models Reference

```python
# Reference: Supported models for suggestions
from plotsense.core.registry_loader import get_registry_loader

registry = get_registry_loader()

print("📋 SUPPORTED MODELS FOR RECOMMENDATIONS:\n")

# OpenAI
print("🤖 OpenAI (Chat):")
for model in registry.get_provider_models("openai", "chat"):
    print(f"   - {model}")

# Groq
print("\n🔥 Groq (Native API):")
for model in registry.get_provider_models("groq", "default"):
    print(f"   - {model}")

# Anthropic
print("\n🔷 Anthropic:")
for model in registry.get_provider_models("anthropic", "default"):
    print(f"   - {model}")
```
